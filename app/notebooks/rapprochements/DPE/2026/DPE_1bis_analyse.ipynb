# Que faire dans certains cas particuliers ?

- Si pas ID-RNB présent initialement, décision simple (on prend le résultat du rappro dès qu'il est unique et >25m2)
- Si ID-RNB d'origine "logiciel", on ne touche pas
- Si ID-RNB d'origine "RNB"
    - si même résultat, on touche pas
    - si résultat différent et 1 seul résultat, on met à jour
    - si résultat différent et N résultat, à investiguer => mettre à jour
    - si résultat différent et 0 résultat, à investiguer => mettre à jour

```python
# chargement des résultats du rapprochement

import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

df_export = pd.read_csv('notebooks/rapprochements/DPE/2026/results_DPE_RNB.csv')
```

# Cas à investiguer : 1 (2025)/ N (2026)

Description de la situation : Le rapprochement 2025 a donné un résultat unique, c'est à dire qu'un ID-RNB a été associé au DPE. En faisant retourner le rapprochement en 2026, nous avons maintenant plusieurs ID-RNB associé au DPE. Que faire ? Mettre à jour ou laisser l'ancien résultat ?

- Il est possible que le résultat ait changé à cause de l'instabilité de clé d'interop, ce qui nous pousserait à ne pas toucher au résultat précédent.
- Il est possible que le résultat ait changé car le RNB est monté en qualité, ce qui nous pousserait à mettre à jour le rapprochement.
- Il est possible que ce problème soit théorique et ne concerne aucun DPE.

## Combien de cas ?

```python
# Nous avions un résultat reprise RNB et maintenant nous en avons plusieurs.
df_1_N = df_export[(df_export['rnb_id_rappro'].apply(len) > 16) & (df_export['provenance_id_rnb'] == 'Reprise RNB')]
```
